## 1. Solución Analítica

Definimos una nueva variable aleatoria $W$ que represente la holgura o ajuste del ensamble mecánico. Esta variable se define como la diferencia entre el diámetro interno del cojinete ($x_1$) y el diámetro de la flecha ($X_2$):

$$W = x_1 - X_2$$

Como $x_1$ y $X_2$ son variables aleatorias independientes que siguen una distribución normal, cualquier combinación lineal de ellas también se distribuirá de forma normal: $W \sim N(\mu_W, \sigma_W^2)$.

 Parámetros de $W$

* **Media de la holgura ($\mu_W$):**
  $$\mu_W = \mu_{x_1} - \mu_{X_2} = 1.5 - 1.48 = 0.02$$

* **Varianza de la holgura ($\sigma_W^2$):** Dado que las variables son independientes, las varianzas se suman:
  $$\sigma_W^2 = \sigma_{x_1}^2 + \sigma_{X_2}^2 = 0.0016 + 0.0009 = 0.0025$$

* **Desviación estándar de la holgura ($\sigma_W$):**
  $$\sigma_W = \sqrt{\sigma_W^2} = \sqrt{0.0025} = 0.05$$

Por lo tanto, la variable de la holgura queda completamente definida como:
$$W \sim N(0.02, 0.0025)$$

### A) Probabilidad de que haya interferencia

Existe interferencia mecánica cuando la flecha es más grande que el espacio disponible en el cojinete, cuando la holgura es negativa $W < 0$.

Para calcular esta probabilidad, vamos a estandarizar la variable $W$ para transformarla en una distribución normal estándar $Z \sim N(0, 1)$ mediante la fórmula $Z = \frac{W - \mu_W}{\sigma_W}$:

$$P(W < 0) = P\left(\frac{W - \mu_W}{\sigma_W} < \frac{0 - 0.02}{0.05}\right)$$

$$P(W < 0) = P\left(Z < \frac{-0.02}{0.05}\right) = P(Z < -0.4)$$

Buscando el valor de $Z = -0.4$ en las tablas de la distribución normal estándar acumulada:

$$P(Z < -0.4) \approx 0.3446$$

**Resultados:** La probabilidad de que ocurra interferencia en el ensamble es de **34.46%**.

### B) Número de experimentos necesarios para la simulación ($N$)

Tomando el Teorema del Límite Central. necesitamos:
* **Error máximo ($e$):** $0.01$
* **Nivel de confianza:** $95\% \implies \alpha = 0.05$
* **Probabilidad calculada ($p$):** $0.3446$

Para un nivel de confianza del $95\%$, buscamos el valor crítico de $Z$ ($\alpha/2 = 0.025$).

$$Z_{0.025} = 1.96$$


La fórmula para $N$ es:

$$N = \frac{Z_{\alpha/2}^2 \cdot p \cdot (1 - p)}{e^2}$$

Sustituyendo los valores conocidos:

$$N = \frac{(1.96)^2 \cdot 0.3446 \cdot (1 - 0.3446)}{(0.01)^2}$$

$$N = \frac{3.8416 \cdot 0.3446 \cdot 0.6554}{0.0001}$$

$$N = \frac{3.8416 \cdot 0.22585}{0.0001} = \frac{0.86758}{0.0001} \approx 8675.8$$


**Resultados:** Es necesario simular el experimento al menos **8,675 veces**.

**Para la solución por simulación:**

```python
import numpy as np
import scipy.stats as stats

# parámetros del problema
mu_x1 = 1.5
varianza_x1 = 0.0016
sigma_x1 = np.sqrt(varianza_x1) # 0.04

mu_x2 = 1.48
varianza_x2 = 0.0009
sigma_x2 = np.sqrt(varianza_x2) # 0.03

# Número de simulaciones para una estimación inicial
N_inicial = 1_000_000

# Generar muestras aleatorias utilizando el método de Montecarlo
np.random.seed(42) # Semilla
muestras_x1 = np.random.normal(mu_x1, sigma_x1, N_inicial)
muestras_x2 = np.random.normal(mu_x2, sigma_x2, N_inicial)

# Condición de interferencia (Flecha > Cojinete)
interferencia = muestras_x2 > muestras_x1

# Probabilidad estimada
prob_estimada = np.mean(interferencia)

print(f"Probabilidad estimada de interferencia (con N={N_inicial}): {prob_estimada:.4f} ({prob_estimada * 100:.2f}%)")

# Parámetros para el inciso b)
error_max = 0.01
confianza = 0.95

# Valor Z para el 95% de confianza
Z = stats.norm.ppf(1 - (1 - confianza) / 2) # Aprox 1.96

# Tomamos la probabilidad estimada como aproximación de 'p'
p = prob_estimada

# Fórmula del tamaño de muestra
N_necesario = (Z**2 * p * (1 - p)) / (error_max**2)
N_necesario = int(np.ceil(N_necesario)) # Redondear hacia arriba

print(f"Número de experimentos necesarios: {N_necesario}")
```
